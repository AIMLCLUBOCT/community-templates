Constructing and training a feed-forward multi-layer perceptron (MLP) in PyTorch.

```python
import torch
import torch.nn as nn
import torch.optim as optim

class SimpleMLP(nn.Module):
    def __init__(self, in_features=10, hidden_dim=32, out_classes=2):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_features, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, out_classes)
        )
    def forward(self, x):
        return self.net(x)

model = SimpleMLP()
print(model)
```
